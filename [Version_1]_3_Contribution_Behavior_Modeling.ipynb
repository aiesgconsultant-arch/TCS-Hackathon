```python
# ============================================================================
# ENHANCED NEST CONTRIBUTION BEHAVIOR PREDICTION - DATA GENERATION
# Improved realism, balance, and business logic
# ============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from datetime import datetime, timedelta
import uuid
import random

# ML imports
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold, GridSearchCV
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, VotingClassifier
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix,
                           roc_auc_score, roc_curve, precision_recall_curve,
                           precision_score, recall_score, f1_score)
from sklearn.feature_selection import SelectKBest, f_classif, RFE, RFECV, mutual_info_classif
from sklearn.utils import resample
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler

warnings.filterwarnings('ignore')
np.random.seed(42)
random.seed(42)

print("=== ENHANCED NEST CONTRIBUTION BEHAVIOR PREDICTION SYSTEM ===")
print()

# ============================================================================
# SECTION 0: ENHANCED SYNTHETIC DATA GENERATION
# ============================================================================

def generate_enhanced_contribution_behavior_data(n_records=10000):
    """Generate realistic and balanced contribution behavior dataset with improved logic"""

    print("Generating enhanced contribution behavior dataset...")

    # Define more comprehensive demographics and realistic distributions
    sectors = ['Public', 'Private_Services', 'Healthcare', 'Education', 'Technology',
              'Finance', 'Manufacturing', 'Retail', 'Construction']

    regions = ['London', 'South_East', 'North_West', 'Scotland', 'Wales',
              'Yorkshire', 'South_West', 'Midlands', 'North_East']

    education_levels = ['Secondary', 'A_Level', 'Undergraduate', 'Postgraduate']

    # More realistic sector salary distributions (UK 2024/25)
    sector_salary_stats = {
        'Finance': {'mean': 58000, 'std': 22000, 'min': 25000, 'max': 150000},
        'Technology': {'mean': 55000, 'std': 20000, 'min': 28000, 'max': 120000},
        'Healthcare': {'mean': 42000, 'std': 15000, 'min': 22000, 'max': 95000},
        'Public': {'mean': 38000, 'std': 12000, 'min': 20000, 'max': 85000},
        'Education': {'mean': 35000, 'std': 10000, 'min': 22000, 'max': 75000},
        'Manufacturing': {'mean': 34000, 'std': 12000, 'min': 20000, 'max': 80000},
        'Private_Services': {'mean': 32000, 'std': 15000, 'min': 18000, 'max': 90000},
        'Retail': {'mean': 28000, 'std': 8000, 'min': 18000, 'max': 60000},
        'Construction': {'mean': 36000, 'std': 14000, 'min': 20000, 'max': 85000}
    }

    # Regional cost of living multipliers
    regional_col_multipliers = {
        'London': 1.4, 'South_East': 1.2, 'South_West': 1.05,
        'Midlands': 0.95, 'Yorkshire': 0.9, 'North_West': 0.9,
        'North_East': 0.85, 'Scotland': 0.95, 'Wales': 0.9
    }

    data = []

    # Pre-calculate target distribution for better balance
    # Realistic UK pension contribution behavior distribution
    target_distribution = {
        'No_Change': 0.50,    # Most people maintain current contributions
        'Increase': 0.25,     # Economic optimism and life stage improvements
        'Decrease': 0.15,     # Financial pressures
        'Pause_Stop': 0.10   # Severe financial stress or life events
    }

    # Generate behavioral categories first to ensure balance
    n_behaviors = {behavior: int(n_records * pct) for behavior, pct in target_distribution.items()}
    # Adjust for rounding
    remaining = n_records - sum(n_behaviors.values())
    n_behaviors['No_Change'] += remaining

    # Create behavior assignments
    behavior_assignments = []
    for behavior, count in n_behaviors.items():
        behavior_assignments.extend([behavior] * count)

    # Shuffle to randomize order
    np.random.shuffle(behavior_assignments)

    for i in range(n_records):
        member_id = f"NEST{str(i+100001).zfill(6)}"
        target_behavior = behavior_assignments[i]

        # Generate demographics with realistic correlations
        age = np.random.normal(42, 13)  # UK working population average
        age = max(18, min(68, age))

        # Gender distribution (slightly more female in NEST due to auto-enrollment patterns)
        gender = np.random.choice(['Male', 'Female'], p=[0.48, 0.52])

        # Education level varies by age (younger more educated)
        if age < 30:
            education_probs = [0.15, 0.25, 0.45, 0.15]
        elif age < 45:
            education_probs = [0.20, 0.30, 0.35, 0.15]
        else:
            education_probs = [0.30, 0.35, 0.25, 0.10]

        education = np.random.choice(education_levels, p=education_probs)

        # Sector influenced by education and age
        if education in ['Undergraduate', 'Postgraduate']:
            sector_probs = [0.08, 0.12, 0.15, 0.15, 0.20, 0.15, 0.05, 0.05, 0.05]
        else:
            sector_probs = [0.15, 0.15, 0.10, 0.08, 0.05, 0.05, 0.20, 0.15, 0.07]

        sector = np.random.choice(sectors, p=sector_probs)
        region = np.random.choice(regions)

        # Generate salary based on multiple realistic factors
        salary_stats = sector_salary_stats[sector]
        base_salary = np.random.normal(salary_stats['mean'], salary_stats['std'])

        # Education multipliers
        education_multipliers = {'Secondary': 0.85, 'A_Level': 0.95,
                               'Undergraduate': 1.15, 'Postgraduate': 1.35}
        base_salary *= education_multipliers[education]

        # Age/experience curve (more realistic)
        experience_years = max(0, age - 18)
        experience_multiplier = 1 + (experience_years * 0.012) - (experience_years * 0.00008) ** 2
        base_salary *= experience_multiplier

        # Gender pay gap (unfortunately realistic)
        if gender == 'Female':
            base_salary *= 0.93

        # Regional adjustment
        base_salary *= regional_col_multipliers[region]

        # Add some randomness and cap within realistic bounds
        base_salary += np.random.normal(0, salary_stats['std'] * 0.2)
        current_salary = max(salary_stats['min'],
                           min(salary_stats['max'], base_salary))

        # Employment characteristics
        # Tenure varies by sector and age
        if sector == 'Public':
            avg_tenure = 48  # Public sector has longer tenure
        elif sector in ['Technology', 'Finance']:
            avg_tenure = 24  # More mobile sectors
        else:
            avg_tenure = 36

        tenure_months = max(1, np.random.exponential(avg_tenure))
        tenure_months = min(tenure_months, (age - 18) * 12)  # Can't work longer than adult life

        # Job changes correlated with sector and age
        job_change_rate = 1.2 if sector in ['Technology', 'Private_Services'] else 0.6
        if age < 30:
            job_change_rate *= 1.5
        elif age > 50:
            job_change_rate *= 0.5

        job_changes_5yr = np.random.poisson(job_change_rate)
        years_with_nest = min(tenure_months / 12, np.random.exponential(3.5))

        # Financial situation with realistic correlations
        # Higher earners more likely to have other investments
        other_pension_prob = min(0.8, 0.1 + (current_salary - 20000) / 100000)
        has_other_pension = np.random.random() < other_pension_prob

        isa_prob = min(0.9, 0.2 + (current_salary - 20000) / 80000)
        has_isa = np.random.random() < isa_prob

        # Homeownership varies by age, salary, and region
        homeowner_prob = min(0.95, 0.3 + (age - 25) * 0.01 + (current_salary - 25000) / 100000)
        if region == 'London':
            homeowner_prob *= 0.7  # London housing crisis
        homeowner = np.random.random() < homeowner_prob

        # Life events with age-appropriate probabilities
        recent_job_change = job_changes_5yr > 0 and np.random.random() < 0.2

        # Salary changes more likely for younger workers
        salary_change_prob = 0.3 if age < 35 else 0.15
        recent_salary_change = np.random.random() < salary_change_prob

        salary_change_pct = 0
        if recent_salary_change:
            # More realistic salary change distribution
            if np.random.random() < 0.6:  # 60% positive changes
                salary_change_pct = np.random.lognormal(np.log(0.05), 0.3)
            else:  # 40% negative changes
                salary_change_pct = -np.random.lognormal(np.log(0.03), 0.2)

        # Family changes vary by age
        if 25 <= age <= 40:
            family_change_prob = 0.15  # Peak family formation years
        elif 40 < age <= 55:
            family_change_prob = 0.08  # Some changes but less frequent
        else:
            family_change_prob = 0.05
        family_change = np.random.random() < family_change_prob

        health_event_prob = 0.03 + (age - 25) * 0.001  # Increases with age
        health_event = np.random.random() < health_event_prob

        # Economic stress indicators with regional variation
        base_col_stress_probs = [0.35, 0.45, 0.20]  # Low, Medium, High
        if region in ['London', 'South_East']:
            base_col_stress_probs = [0.25, 0.50, 0.25]  # Higher stress in expensive areas

        cost_of_living_stress = np.random.choice(['Low', 'Medium', 'High'],
                                               p=base_col_stress_probs)

        # Debt to income varies by age and salary
        if age < 30:
            dti_mean = 0.35  # Young people often have more debt
        elif current_salary > 50000:
            dti_mean = 0.20  # Higher earners manage debt better
        else:
            dti_mean = 0.28

        debt_to_income = max(0, np.random.normal(dti_mean, 0.15))
        debt_to_income = min(0.8, debt_to_income)  # Cap at 80%

        # Calculate disposable income more realistically
        # Essential costs vary by region and family situation
        base_essential_pct = 0.65

        if region in ['London', 'South_East']:
            base_essential_pct += 0.1
        if family_change:  # Recent family changes increase costs
            base_essential_pct += 0.05
        if cost_of_living_stress == 'High':
            base_essential_pct = 0.85
        elif cost_of_living_stress == 'Medium':
            base_essential_pct = 0.75

        disposable_income = current_salary * (1 - base_essential_pct - debt_to_income)
        disposable_income = max(0, disposable_income)

        # Historical contribution patterns - more realistic baseline
        sector_contrib_means = {
            'Public': 0.09, 'Finance': 0.085, 'Technology': 0.08,
            'Healthcare': 0.075, 'Education': 0.08, 'Manufacturing': 0.07,
            'Private_Services': 0.065, 'Retail': 0.055, 'Construction': 0.06
        }

        base_contrib = sector_contrib_means.get(sector, 0.07)
        if education in ['Undergraduate', 'Postgraduate']:
            base_contrib *= 1.1
        if age > 45:
            base_contrib *= 1.15  # Pre-retirement awareness

        current_contribution_pct = max(0.01, np.random.normal(base_contrib, 0.025))
        monthly_contribution = current_salary / 12 * current_contribution_pct

        # ENHANCED: Engagement metrics with realistic correlations
        # Engagement correlates with education, age, and contribution behavior
        base_engagement = 20

        if education in ['Undergraduate', 'Postgraduate']:
            base_engagement += 15
        if age > 35:
            base_engagement += 10  # Older workers more engaged with retirement
        if current_contribution_pct > 0.08:
            base_engagement += 20  # High contributors more engaged

        # Individual engagement activities
        app_logins_6m = max(0, np.random.poisson(base_engagement * 0.4))
        viewed_balance_6m = max(0, np.random.poisson(base_engagement * 0.6))
        opened_emails_6m = max(0, np.random.poisson(base_engagement * 0.3))
        contacted_support_6m = np.random.poisson(max(0.3, base_engagement * 0.025))

        engagement_score = min(100,
                              app_logins_6m * 2.5 +
                              viewed_balance_6m * 1.8 +
                              opened_emails_6m * 3.2 +
                              contacted_support_6m * 4.5)

        # Pension knowledge with realistic distributions
        calc_prob = 0.15 + (education in ['Undergraduate', 'Postgraduate']) * 0.2 + (age > 40) * 0.15
        completed_pension_calc = np.random.random() < calc_prob

        webinar_prob = 0.05 + (education == 'Postgraduate') * 0.1 + (age > 45) * 0.08
        attended_webinar = np.random.random() < webinar_prob

        # Knowledge score based on education and engagement
        knowledge_base = 25
        if education == 'Postgraduate':
            knowledge_base = 45
        elif education == 'Undergraduate':
            knowledge_base = 35
        elif education == 'A_Level':
            knowledge_base = 30

        knowledge_score = knowledge_base
        if completed_pension_calc:
            knowledge_score += 30
        if attended_webinar:
            knowledge_score += 25
        if engagement_score > 50:
            knowledge_score += 15

        knowledge_score = min(100, knowledge_score + np.random.randint(-10, 15))

        # External economic factors
        economic_outlook = np.random.choice(['Optimistic', 'Neutral', 'Pessimistic'],
                                          p=[0.20, 0.55, 0.25])
        inflation_impact = np.random.normal(0.06, 0.02)  # 2024/25 inflation reality

        # ENHANCED: Behavior determination with sophisticated scoring
        # Start with target behavior and work backwards to create realistic correlations

        behavior_score = 0

        # Adjust base factors to align with target behavior
        if target_behavior == 'Increase':
            # Bias factors positively for increase behavior
            if disposable_income < 3000:
                disposable_income *= 1.3  # Boost disposable income
            if engagement_score < 40:
                engagement_score = min(100, engagement_score * 1.4)
            if economic_outlook == 'Pessimistic':
                economic_outlook = np.random.choice(['Neutral', 'Optimistic'])

        elif target_behavior == 'Pause_Stop':
            # Bias factors negatively for pause/stop behavior
            if cost_of_living_stress == 'Low':
                cost_of_living_stress = np.random.choice(['Medium', 'High'])
            if debt_to_income < 0.3:
                debt_to_income = min(0.6, debt_to_income * 1.5)
            if not recent_job_change and np.random.random() < 0.3:
                recent_job_change = True

        elif target_behavior == 'Decrease':
            # Moderate negative bias for decrease behavior
            if cost_of_living_stress == 'Low' and np.random.random() < 0.5:
                cost_of_living_stress = 'Medium'
            if disposable_income > 4000:
                disposable_income *= 0.8

        # Now calculate behavior score normally
        # Positive factors
        if disposable_income > 6000:
            behavior_score += 35
        elif disposable_income > 3000:
            behavior_score += 20
        elif disposable_income > 1500:
            behavior_score += 10

        if engagement_score > 60:
            behavior_score += 25
        elif engagement_score > 35:
            behavior_score += 15
        elif engagement_score > 15:
            behavior_score += 8

        if knowledge_score > 70:
            behavior_score += 20
        elif knowledge_score > 45:
            behavior_score += 12
        elif knowledge_score > 25:
            behavior_score += 6

        if years_with_nest > 4:
            behavior_score += 12
        elif years_with_nest > 2:
            behavior_score += 6

        if economic_outlook == 'Optimistic':
            behavior_score += 18
        elif economic_outlook == 'Pessimistic':
            behavior_score -= 25

        if recent_salary_change and salary_change_pct > 0.08:
            behavior_score += 30
        elif recent_salary_change and salary_change_pct > 0.03:
            behavior_score += 15
        elif recent_salary_change and salary_change_pct < -0.08:
            behavior_score -= 35
        elif recent_salary_change and salary_change_pct < -0.03:
            behavior_score -= 20

        # Negative factors
        if cost_of_living_stress == 'High':
            behavior_score -= 30
        elif cost_of_living_stress == 'Medium':
            behavior_score -= 15

        if debt_to_income > 0.45:
            behavior_score -= 25
        elif debt_to_income > 0.35:
            behavior_score -= 15
        elif debt_to_income > 0.25:
            behavior_score -= 8

        if recent_job_change:
            behavior_score -= 18

        if family_change:
            behavior_score -= 12

        if health_event:
            behavior_score -= 22

        # Age and life stage factors
        if age < 25:
            behavior_score -= 15  # Young adults financial instability
        elif 25 <= age < 35:
            behavior_score -= 5   # Early career challenges
        elif 45 <= age < 55:
            behavior_score += 8   # Peak earning years
        elif age >= 55:
            behavior_score += 15  # Pre-retirement focus

        # Sector-specific factors
        if sector in ['Public', 'Healthcare', 'Education']:
            behavior_score += 5  # More stable employment
        elif sector in ['Technology', 'Finance']:
            behavior_score += 3  # Higher earning potential
        elif sector in ['Retail', 'Construction']:
            behavior_score -= 5  # More volatile employment

        # Add controlled randomness (reduced from original)
        behavior_score += np.random.normal(0, 8)

        # Convert score to behavior - but use target behavior with some flexibility
        if np.random.random() < 0.85:  # 85% of time use target behavior
            behavior = target_behavior
        else:  # 15% of time use score-based behavior for realism
            if behavior_score > 25:
                behavior = 'Increase'
            elif behavior_score < -15:
                behavior = 'Pause_Stop'
            elif behavior_score < 5:
                behavior = 'Decrease'
            else:
                behavior = 'No_Change'

        # Create record with all enhanced features
        record = {
            'member_id': member_id,
            'age': round(age, 1),
            'gender': gender,
            'education_level': education,
            'employment_sector': sector,
            'region': region,
            'current_salary': round(current_salary, 2),
            'tenure_months': round(tenure_months, 1),
            'years_with_nest': round(years_with_nest, 2),
            'job_changes_5yr': job_changes_5yr,
            'current_contribution_pct': round(current_contribution_pct, 4),
            'monthly_contribution': round(monthly_contribution, 2),
            'has_other_pension': has_other_pension,
            'has_isa': has_isa,
            'homeowner': homeowner,
            'recent_job_change': recent_job_change,
            'recent_salary_change': recent_salary_change,
            'salary_change_pct': round(salary_change_pct, 4),
            'family_change': family_change,
            'health_event': health_event,
            'cost_of_living_stress': cost_of_living_stress,
            'debt_to_income': round(debt_to_income, 3),
            'disposable_income': round(disposable_income, 2),
            'app_logins_6m': app_logins_6m,
            'viewed_balance_6m': viewed_balance_6m,
            'opened_emails_6m': opened_emails_6m,
            'contacted_support_6m': contacted_support_6m,
            'engagement_score': round(engagement_score, 1),
            'completed_pension_calc': completed_pension_calc,
            'attended_webinar': attended_webinar,
            'knowledge_score': knowledge_score,
            'economic_outlook': economic_outlook,
            'inflation_impact': round(inflation_impact, 4),
            'behavior_score': round(behavior_score, 2),  # Added for analysis
            'contribution_behavior': behavior  # TARGET VARIABLE
        }

        data.append(record)

    return pd.DataFrame(data)

# Generate enhanced dataset
print("Generating enhanced dataset with improved realism and balance...")
df_enhanced = generate_enhanced_contribution_behavior_data(10000)

# Save to CSV
csv_filename = 'nest_enhanced_contribution_behavior_dataset.csv'
df_enhanced.to_csv(csv_filename, index=False)
print(f"Enhanced dataset saved as '{csv_filename}'")

# Display comprehensive statistics
print(f"Dataset shape: {df_enhanced.shape}")
print(f"\n=== TARGET VARIABLE DISTRIBUTION ===")
behavior_counts = df_enhanced['contribution_behavior'].value_counts()
print(behavior_counts)
print(f"\nBehavior percentages:")
for behavior, count in behavior_counts.items():
    percentage = (count / len(df_enhanced)) * 100
    print(f"{behavior}: {count:,} members ({percentage:.1f}%)")

print(f"\n=== DEMOGRAPHIC INSIGHTS ===")
print(f"Age distribution: {df_enhanced['age'].describe()}")
print(f"\nSalary distribution: £{df_enhanced['current_salary'].describe()}")
print(f"\nSector distribution:")
print(df_enhanced['employment_sector'].value_counts())

print(f"\n=== ENGAGEMENT METRICS ===")
print(f"Engagement score: {df_enhanced['engagement_score'].describe()}")
print(f"Knowledge score: {df_enhanced['knowledge_score'].describe()}")

print(f"\n=== FINANCIAL STRESS INDICATORS ===")
print(f"Cost of living stress: {df_enhanced['cost_of_living_stress'].value_counts()}")
print(f"Debt to income ratio: {df_enhanced['debt_to_income'].describe()}")

print(f"\n=== SAMPLE DATA ===")
sample_cols = ['age', 'employment_sector', 'current_salary', 'engagement_score',
               'disposable_income', 'cost_of_living_stress', 'behavior_score', 'contribution_behavior']
print(df_enhanced[sample_cols].head(10))

print(f"\n=== DATA QUALITY CHECKS ===")
print(f"Missing values: {df_enhanced.isnull().sum().sum()}")
print(f"Duplicated member IDs: {df_enhanced['member_id'].duplicated().sum()}")
print(f"Salary range: £{df_enhanced['current_salary'].min():,.0f} - £{df_enhanced['current_salary'].max():,.0f}")
print(f"Age range: {df_enhanced['age'].min():.1f} - {df_enhanced['age'].max():.1f} years")

# Correlation analysis between behavior score and actual behavior
behavior_mapping = {'Increase': 3, 'No_Change': 2, 'Decrease': 1, 'Pause_Stop': 0}
df_enhanced['behavior_numeric'] = df_enhanced['contribution_behavior'].map(behavior_mapping)
correlation = df_enhanced['behavior_score'].corr(df_enhanced['behavior_numeric'])
print(f"\nBehavior score correlation with target: {correlation:.3f}")

print(f"\n" + "="*60)
print("ENHANCED DATASET GENERATION COMPLETE")
print("Key Improvements:")
print("✓ More realistic salary distributions by sector")
print("✓ Regional cost of living adjustments")
print("✓ Age-appropriate demographic correlations")
print("✓ Improved target variable balance")
print("✓ Enhanced engagement and knowledge scoring")
print("✓ Sophisticated behavior determination logic")
print("✓ Added behavior score for model interpretability")
print("="*60)
```

```python
# ============================================================================
# SECTION 3: FEATURE ENGINEERING AND ENCODING
# ============================================================================

print("\n" + "="*60)
print("SECTION 3: FEATURE ENGINEERING AND ENCODING")
print("="*60)

# Create a copy for processing
df_processed = contribution_data.copy()

# Remove ID column
df_processed = df_processed.drop('member_id', axis=1)

# Feature engineering
print("Creating new features...")

# 1. Financial stress indicator
df_processed['financial_stress'] = (
    (df_processed['debt_to_income'] > 0.3).astype(int) +
    (df_processed['cost_of_living_stress'] == 'High').astype(int) +
    (df_processed['disposable_income'] < 2000).astype(int)
)

# 2. Career stability score
df_processed['career_stability'] = (
    (df_processed['tenure_months'] > 36).astype(int) +
    (df_processed['job_changes_5yr'] <= 1).astype(int) +
    (~df_processed['recent_job_change']).astype(int)
)

# 3. Pension engagement level
df_processed['pension_engagement'] = (
    (df_processed['engagement_score'] > 30).astype(int) +
    df_processed['completed_pension_calc'].astype(int) +
    df_processed['attended_webinar'].astype(int) +
    (df_processed['knowledge_score'] > 50).astype(int)
)

# 4. Life stage indicator
def get_life_stage(age):
    if age < 30:
        return 'Young_Professional'
    elif age < 45:
        return 'Mid_Career'
    elif age < 55:
        return 'Senior_Career'
    else:
        return 'Pre_Retirement'

df_processed['life_stage'] = df_processed['age'].apply(get_life_stage)

# 5. Income quintile
df_processed['income_quintile'] = pd.qcut(df_processed['current_salary'],
                                         q=5,
                                         labels=['Q1_Lowest', 'Q2_Low', 'Q3_Medium', 'Q4_High', 'Q5_Highest'])

print(f"New features created:")
print(f"- financial_stress: {df_processed['financial_stress'].value_counts().to_dict()}")
print(f"- career_stability: {df_processed['career_stability'].value_counts().to_dict()}")
print(f"- pension_engagement: {df_processed['pension_engagement'].value_counts().to_dict()}")
print(f"- life_stage: {df_processed['life_stage'].value_counts().to_dict()}")

# Encode categorical variables
print("\nEncoding categorical variables...")

# Binary encoding for boolean variables
boolean_columns = ['has_other_pension', 'has_isa', 'homeowner', 'recent_job_change',
                  'recent_salary_change', 'family_change', 'health_event',
                  'completed_pension_calc', 'attended_webinar']

for col in boolean_columns:
    df_processed[col] = df_processed[col].astype(int)

# Label encoding for ordinal variables
ordinal_mappings = {
    'education_level': {'Secondary': 1, 'A-Level': 2, 'Undergraduate': 3, 'Postgraduate': 4},
    'cost_of_living_stress': {'Low': 1, 'Medium': 2, 'High': 3},
    'economic_outlook': {'Pessimistic': 1, 'Neutral': 2, 'Optimistic': 3}
}

for col, mapping in ordinal_mappings.items():
    df_processed[col + '_encoded'] = df_processed[col].map(mapping)

# One-hot encoding for nominal variables
nominal_columns = ['gender', 'employment_sector', 'region', 'life_stage', 'income_quintile']

for col in nominal_columns:
    encoded = pd.get_dummies(df_processed[col], prefix=col)
    df_processed = pd.concat([df_processed, encoded], axis=1)

# Encode target variable
label_encoder_target = LabelEncoder()
df_processed['behavior_encoded'] = label_encoder_target.fit_transform(df_processed['contribution_behavior'])

# Create mapping dictionary for reference
behavior_mapping = dict(zip(label_encoder_target.classes_,
                           label_encoder_target.transform(label_encoder_target.classes_)))
print(f"\nTarget variable encoding: {behavior_mapping}")

print(f"Dataset shape after encoding: {df_processed.shape}")
```

```python
# ============================================================================
# SECTION 4: ENHANCED FEATURE SELECTION
# ============================================================================

print("\n" + "="*60)
print("SECTION 4: ENHANCED FEATURE SELECTION")
print("="*60)

# Select features for modeling (exclude original categorical columns and target)
exclude_columns = ['contribution_behavior'] + nominal_columns + list(ordinal_mappings.keys())
feature_columns = [col for col in df_processed.columns
                  if col not in exclude_columns and col != 'behavior_encoded']

X = df_processed[feature_columns]
y = df_processed['behavior_encoded']

print(f"Initial number of features for modeling: {len(feature_columns)}")
print(f"Features: {feature_columns[:10]}..." if len(feature_columns) > 10 else f"Features: {feature_columns}")

# Comprehensive data cleaning for feature selection
print("\nPreparing features for selection...")

# 1. Check for any remaining non-numeric columns
non_numeric = X.select_dtypes(exclude=[np.number]).columns
if len(non_numeric) > 0:
    print(f"Non-numeric columns found: {list(non_numeric)}")

    # Try to convert non-numeric columns to numeric
    for col in non_numeric:
        if X[col].dtype == 'object':
            # Try to convert to numeric, coerce errors to NaN
            X[col] = pd.to_numeric(X[col], errors='coerce')
            print(f"  Converted {col} to numeric")

    # After conversion, select only numeric columns
    X = X.select_dtypes(include=[np.number])
    print(f"After conversion, kept {X.shape[1]} numeric columns")

# 2. Handle any NaN values
print(f"\nChecking for NaN values...")
nan_counts = X.isnull().sum()
total_nans = nan_counts.sum()

if total_nans > 0:
    print(f"Found NaN values in {total_nans} cells across {(nan_counts > 0).sum()} columns")
    print("Columns with NaN values:")
    for col in nan_counts[nan_counts > 0].index:
        print(f"  {col}: {nan_counts[col]} NaN values")

    # Fill NaN values appropriately
    for col in X.columns:
        if X[col].isnull().sum() > 0:
            if col in ['age', 'current_salary', 'tenure_months', 'years_with_nest',
                      'engagement_score', 'knowledge_score', 'disposable_income']:
                # Use median for continuous variables
                fill_value = X[col].median()
                X[col] = X[col].fillna(fill_value)
                print(f"  Filled {col} NaN values with median: {fill_value:.2f}")
            else:
                # Use mode (most frequent) for categorical/binary encoded variables
                if len(X[col].mode()) > 0:
                    fill_value = X[col].mode()[0]
                else:
                    fill_value = 0
                X[col] = X[col].fillna(fill_value)
                print(f"  Filled {col} NaN values with mode/default: {fill_value}")
else:
    print("No NaN values found")

# 3. Check for infinite values
print(f"\nChecking for infinite values...")
inf_mask = np.isinf(X)
inf_counts = inf_mask.sum().sum()

if inf_counts > 0:
    print(f"Found {inf_counts} infinite values")
    inf_cols = inf_mask.sum()
    print("Columns with infinite values:")
    for col in inf_cols[inf_cols > 0].index:
        print(f"  {col}: {inf_cols[col]} infinite values")

    # Replace infinite values with NaN, then fill with median/mode
    X = X.replace([np.inf, -np.inf], np.nan)

    # Fill the new NaN values
    for col in X.columns:
        if X[col].isnull().sum() > 0:
            fill_value = X[col].median()
            X[col] = X[col].fillna(fill_value)
            print(f"  Replaced infinite values in {col} with median: {fill_value:.2f}")
else:
    print("No infinite values found")

# 4. Ensure all features are numeric and valid
print(f"\nFinal data validation...")
print(f"Shape: {X.shape}")
print(f"All columns numeric: {X.dtypes.apply(lambda x: pd.api.types.is_numeric_dtype(x)).all()}")
print(f"No missing values: {X.isnull().sum().sum() == 0}")
print(f"No infinite values: {np.isinf(X).sum().sum() == 0}")

# Update feature list after cleaning
feature_columns = X.columns.tolist()
print(f"Final number of clean features: {len(feature_columns)}")

# Verify target variable
print(f"\nTarget variable check:")
print(f"Target shape: {y.shape}")
print(f"Target distribution: {pd.Series(y).value_counts().sort_index().to_dict()}")
print(f"Target has no missing values: {y.isnull().sum() == 0}")

# Feature selection using multiple methods
print("\nApplying feature selection...")

# Method 1: Statistical selection (ANOVA F-test)
try:
    k_stats = min(20, len(feature_columns))  # Don't select more features than available
    selector_stats = SelectKBest(score_func=f_classif, k=k_stats)
    X_selected_stats = selector_stats.fit_transform(X, y)
    selected_features_stats = X.columns[selector_stats.get_support()].tolist()

    # Get feature scores for interpretation
    feature_scores_stats = pd.DataFrame({
        'feature': X.columns,
        'score': selector_stats.scores_,
        'selected': selector_stats.get_support()
    }).sort_values('score', ascending=False)

    print(f"Statistical feature selection: {len(selected_features_stats)} features selected")
    print("Top 10 features by F-statistic:")
    print(feature_scores_stats.head(10)[['feature', 'score', 'selected']].to_string(index=False))

except Exception as e:
    print(f"Statistical selection failed: {e}")
    selected_features_stats = feature_columns[:20]  # Fallback to first 20 features
    print("Using fallback: first 20 features")

# Method 2: Mutual information
try:
    k_mi = min(20, len(feature_columns))  # Don't select more features than available
    selector_mi = SelectKBest(score_func=mutual_info_classif, k=k_mi)
    X_selected_mi = selector_mi.fit_transform(X, y)
    selected_features_mi = X.columns[selector_mi.get_support()].tolist()

    # Get feature scores for interpretation
    feature_scores_mi = pd.DataFrame({
        'feature': X.columns,
        'score': selector_mi.scores_,
        'selected': selector_mi.get_support()
    }).sort_values('score', ascending=False)

    print(f"\nMutual information selection: {len(selected_features_mi)} features selected")
    print("Top 10 features by mutual information:")
    print(feature_scores_mi.head(10)[['feature', 'score', 'selected']].to_string(index=False))

except Exception as e:
    print(f"Mutual information selection failed: {e}")
    selected_features_mi = feature_columns[:20]  # Fallback to first 20 features
    print("Using fallback: first 20 features")

# Combine selections and create final feature set
print(f"\nCombining feature selections...")
final_features = list(set(selected_features_stats + selected_features_mi))
X_final = X[final_features]

print(f"Final feature set: {len(final_features)} features")
print(f"Features selected by both methods: {len(set(selected_features_stats) & set(selected_features_mi))}")
print(f"Features selected by either method: {len(final_features)}")

# Display final feature list
print(f"\nFinal selected features:")
for i, feature in enumerate(sorted(final_features), 1):
    in_stats = feature in selected_features_stats
    in_mi = feature in selected_features_mi
    print(f"{i:2d}. {feature:<30} [F-test: {'✓' if in_stats else '✗'}] [MI: {'✓' if in_mi else '✗'}]")

# Final verification
print(f"\nFinal dataset verification:")
print(f"X_final shape: {X_final.shape}")
print(f"y shape: {y.shape}")
print(f"All features numeric: {X_final.dtypes.apply(lambda x: pd.api.types.is_numeric_dtype(x)).all()}")
print(f"No missing values in features: {X_final.isnull().sum().sum() == 0}")
print(f"No missing values in target: {y.isnull().sum() == 0}")
print(f"Ready for model training: {X_final.shape[0] == len(y)}")

print("\n" + "="*60)
print("FEATURE SELECTION COMPLETED SUCCESSFULLY")
print("="*60)
```

```python
# ============================================================================
# SECTION 5: MODEL TRAINING AND EVALUATION
# ============================================================================

print("\n" + "="*60)
print("SECTION 5: MODEL TRAINING AND EVALUATION")
print("="*60)

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X_final, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Training set: {X_train.shape}, Test set: {X_test.shape}")
print(f"Class distribution in training set:")
train_dist = pd.Series(y_train).value_counts().sort_index()
for class_idx, count in train_dist.items():
    class_name = label_encoder_target.inverse_transform([class_idx])[0]
    print(f"  {class_name}: {count} ({count/len(y_train)*100:.1f}%)")

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# ============================================================================
# MODEL 1: LOGISTIC REGRESSION
# ============================================================================

print("\n" + "-"*40)
print("MODEL 1: LOGISTIC REGRESSION")
print("-"*40)

# Handle class imbalance with SMOTE
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train_scaled, y_train)

print(f"After SMOTE - Training set: {X_train_balanced.shape}")
balanced_dist = pd.Series(y_train_balanced).value_counts().sort_index()
for class_idx, count in balanced_dist.items():
    class_name = label_encoder_target.inverse_transform([class_idx])[0]
    print(f"  {class_name}: {count}")

# Logistic Regression with cross-validation for C parameter
param_grid_lr = {'C': [0.01, 0.1, 1, 10, 100]}
lr_grid = GridSearchCV(
    LogisticRegression(random_state=42, max_iter=1000, multi_class='ovr'),
    param_grid_lr,
    cv=5,
    scoring='f1_weighted',
    n_jobs=-1
)

lr_grid.fit(X_train_balanced, y_train_balanced)
best_lr = lr_grid.best_estimator_

print(f"Best Logistic Regression parameters: {lr_grid.best_params_}")
print(f"Best cross-validation score: {lr_grid.best_score_:.4f}")

# Predictions
y_pred_lr = best_lr.predict(X_test_scaled)
y_proba_lr = best_lr.predict_proba(X_test_scaled)

# Evaluation
accuracy_lr = accuracy_score(y_test, y_pred_lr)
print(f"\nLogistic Regression Results:")
print(f"Accuracy: {accuracy_lr:.4f}")

print(f"\nClassification Report:")
class_names = label_encoder_target.classes_
print(classification_report(y_test, y_pred_lr, target_names=class_names))

print(f"\nConfusion Matrix:")
cm_lr = confusion_matrix(y_test, y_pred_lr)
print(cm_lr)
```

```python
# ============================================================================
# MODEL 2: RANDOM FOREST
# ============================================================================

print("\n" + "-"*40)
print("MODEL 2: RANDOM FOREST")
print("-"*40)

# Random Forest with hyperparameter tuning
param_grid_rf = {
    'n_estimators': [100, 200],
    'max_depth': [10, 15, None],
    'min_samples_split': [2, 5],
    'class_weight': ['balanced']
}

rf_grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid_rf,
    cv=3,  # Reduced for faster computation
    scoring='f1_weighted',
    n_jobs=-1
)

rf_grid.fit(X_train_scaled, y_train)  # Use original imbalanced data with class_weight='balanced'
best_rf = rf_grid.best_estimator_

print(f"Best Random Forest parameters: {rf_grid.best_params_}")
print(f"Best cross-validation score: {rf_grid.best_score_:.4f}")

# Predictions
y_pred_rf = best_rf.predict(X_test_scaled)
y_proba_rf = best_rf.predict_proba(X_test_scaled)

# Evaluation
accuracy_rf = accuracy_score(y_test, y_pred_rf)
print(f"\nRandom Forest Results:")
print(f"Accuracy: {accuracy_rf:.4f}")

print(f"\nClassification Report:")
print(classification_report(y_test, y_pred_rf, target_names=class_names))

print(f"\nConfusion Matrix:")
cm_rf = confusion_matrix(y_test, y_pred_rf)
print(cm_rf)

# Feature importance
feature_importance = pd.DataFrame({
    'feature': final_features,
    'importance': best_rf.feature_importances_
}).sort_values('importance', ascending=False)

print(f"\nTop 10 Most Important Features:")
print(feature_importance.head(10).to_string(index=False))
```

```python
# ============================================================================
# MODEL 3: GRADIENT BOOSTING
# ============================================================================

print("\n" + "-"*40)
print("MODEL 3: GRADIENT BOOSTING")
print("-"*40)

# Gradient Boosting
gb_model = GradientBoostingClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    random_state=42
)

gb_model.fit(X_train_scaled, y_train)

# Predictions
y_pred_gb = gb_model.predict(X_test_scaled)
y_proba_gb = gb_model.predict_proba(X_test_scaled)

# Evaluation
accuracy_gb = accuracy_score(y_test, y_pred_gb)
print(f"Gradient Boosting Results:")
print(f"Accuracy: {accuracy_gb:.4f}")

print(f"\nClassification Report:")
print(classification_report(y_test, y_pred_gb, target_names=class_names))

print(f"\nConfusion Matrix:")
cm_gb = confusion_matrix(y_test, y_pred_gb)
print(cm_gb)

# ============================================================================
# MODEL 4: ENSEMBLE VOTING
# ============================================================================

print("\n" + "-"*40)
print("MODEL 4: ENSEMBLE VOTING")
print("-"*40)

# Create voting classifier
voting_clf = VotingClassifier(
    estimators=[
        ('lr', best_lr),
        ('rf', best_rf),
        ('gb', gb_model)
    ],
    voting='soft'
)

# For ensemble, use balanced data for logistic regression component
voting_clf.fit(X_train_scaled, y_train)

# Predictions
y_pred_ensemble = voting_clf.predict(X_test_scaled)
y_proba_ensemble = voting_clf.predict_proba(X_test_scaled)

# Evaluation
accuracy_ensemble = accuracy_score(y_test, y_pred_ensemble)
print(f"Ensemble Results:")
print(f"Accuracy: {accuracy_ensemble:.4f}")

print(f"\nClassification Report:")
print(classification_report(y_test, y_pred_ensemble, target_names=class_names))

print(f"\nConfusion Matrix:")
cm_ensemble = confusion_matrix(y_test, y_pred_ensemble)
print(cm_ensemble)
```

```python
# ============================================================================
# SECTION 6: MODEL COMPARISON AND VALIDATION
# ============================================================================

print("\n" + "="*60)
print("SECTION 6: MODEL COMPARISON AND VALIDATION")
print("="*60)

# Compare all models
models_comparison = {
    'Logistic Regression': {
        'accuracy': accuracy_lr,
        'predictions': y_pred_lr,
        'probabilities': y_proba_lr
    },
    'Random Forest': {
        'accuracy': accuracy_rf,
        'predictions': y_pred_rf,
        'probabilities': y_proba_rf
    },
    'Gradient Boosting': {
        'accuracy': accuracy_gb,
        'predictions': y_pred_gb,
        'probabilities': y_proba_gb
    },
    'Ensemble': {
        'accuracy': accuracy_ensemble,
        'predictions': y_pred_ensemble,
        'probabilities': y_proba_ensemble
    }
}

# Create comparison DataFrame
comparison_data = []
for model_name, results in models_comparison.items():
    f1_weighted = f1_score(y_test, results['predictions'], average='weighted')

    comparison_data.append({
        'Model': model_name,
        'Accuracy': results['accuracy'],
        'F1_Weighted': f1_weighted
    })

comparison_df = pd.DataFrame(comparison_data)
print("Model Performance Comparison:")
print(comparison_df.round(4))

# Cross-validation for best model
best_model_name = comparison_df.loc[comparison_df['Accuracy'].idxmax(), 'Model']
print(f"\nBest performing model: {best_model_name}")

if best_model_name == 'Logistic Regression':
    best_model = best_lr
    X_train_for_cv = X_train_balanced
    y_train_for_cv = y_train_balanced
elif best_model_name == 'Random Forest':
    best_model = best_rf
    X_train_for_cv = X_train_scaled
    y_train_for_cv = y_train
elif best_model_name == 'Gradient Boosting':
    best_model = gb_model
    X_train_for_cv = X_train_scaled
    y_train_for_cv = y_train
else:
    best_model = voting_clf
    X_train_for_cv = X_train_scaled
    y_train_for_cv = y_train

# Cross-validation
cv_scores = cross_val_score(best_model, X_train_for_cv, y_train_for_cv,
                           cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
                           scoring='f1_weighted')

print(f"\nCross-validation scores for {best_model_name}:")
print(f"CV F1 scores: {cv_scores}")
print(f"Mean CV F1: {cv_scores.mean():.4f} (+/- {cv_scores.std() * 2:.4f})")

# ============================================================================
```
