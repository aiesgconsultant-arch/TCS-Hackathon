```python
# ============================================================================
# BUSINESS-OPTIMIZED NEST CONTRIBUTION BEHAVIOR PREDICTION - DATA GENERATION
# Enhanced for better predictive power and business actionability
# ============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from datetime import datetime, timedelta
import uuid
import random

# ML imports
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold, GridSearchCV
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, VotingClassifier
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix,
                           roc_auc_score, roc_curve, precision_recall_curve,
                           precision_score, recall_score, f1_score)
from sklearn.feature_selection import SelectKBest, f_classif, RFE, RFECV, mutual_info_classif
from sklearn.utils import resample
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler

warnings.filterwarnings('ignore')
np.random.seed(42)
random.seed(42)

print("=== BUSINESS-OPTIMIZED NEST CONTRIBUTION BEHAVIOR PREDICTION ===")
print()

# ============================================================================
# SECTION 0: BUSINESS-OPTIMIZED DATA GENERATION
# ============================================================================

def generate_business_optimized_data(n_records=12000):
    """
    Generate dataset optimized for business results with:
    1. Better class balance for actionable insights
    2. Stronger predictive signals
    3. More realistic intervention scenarios
    4. Enhanced feature engineering opportunities
    """

    print("Generating business-optimized contribution behavior dataset...")

    # IMPROVED: More balanced target distribution for better model performance
    # Focus on actionable business scenarios
    target_distribution = {
        'No_Change': 0.40,      # Reduced from 50% - stable baseline
        'Increase': 0.30,       # Increased - more growth opportunities
        'Decrease': 0.20,       # Increased - critical early warning signals
        'Pause_Stop': 0.10      # Maintained - highest priority interventions
    }

    print(f"Target distribution: {target_distribution}")

    # Enhanced demographic and sector definitions
    sectors = ['Public', 'Private_Services', 'Healthcare', 'Education', 'Technology',
              'Finance', 'Manufacturing', 'Retail', 'Construction', 'Hospitality']

    regions = ['London', 'South_East', 'North_West', 'Scotland', 'Wales',
              'Yorkshire', 'South_West', 'Midlands', 'North_East', 'Northern_Ireland']

    education_levels = ['Secondary', 'A_Level', 'Undergraduate', 'Postgraduate', 'Professional']

    # BUSINESS-CRITICAL: Enhanced salary distributions with more variation
    sector_salary_stats = {
        'Finance': {'mean': 62000, 'std': 28000, 'min': 28000, 'max': 180000},
        'Technology': {'mean': 58000, 'std': 24000, 'min': 32000, 'max': 140000},
        'Healthcare': {'mean': 44000, 'std': 18000, 'min': 24000, 'max': 110000},
        'Public': {'mean': 41000, 'std': 15000, 'min': 22000, 'max': 95000},
        'Education': {'mean': 38000, 'std': 12000, 'min': 24000, 'max': 85000},
        'Manufacturing': {'mean': 37000, 'std': 16000, 'min': 22000, 'max': 90000},
        'Private_Services': {'mean': 35000, 'std': 18000, 'min': 20000, 'max': 100000},
        'Construction': {'mean': 39000, 'std': 17000, 'min': 22000, 'max': 95000},
        'Retail': {'mean': 29000, 'std': 10000, 'min': 19000, 'max': 70000},
        'Hospitality': {'mean': 26000, 'std': 8000, 'min': 18000, 'max': 55000}
    }

    # Enhanced regional variations
    regional_factors = {
        'London': {'col_multiplier': 1.45, 'salary_premium': 1.25, 'stress_factor': 1.3},
        'South_East': {'col_multiplier': 1.25, 'salary_premium': 1.15, 'stress_factor': 1.1},
        'South_West': {'col_multiplier': 1.08, 'salary_premium': 1.05, 'stress_factor': 0.9},
        'Midlands': {'col_multiplier': 0.95, 'salary_premium': 0.98, 'stress_factor': 0.95},
        'Yorkshire': {'col_multiplier': 0.92, 'salary_premium': 0.95, 'stress_factor': 0.9},
        'North_West': {'col_multiplier': 0.93, 'salary_premium': 0.96, 'stress_factor': 0.9},
        'North_East': {'col_multiplier': 0.88, 'salary_premium': 0.92, 'stress_factor': 0.85},
        'Scotland': {'col_multiplier': 0.97, 'salary_premium': 0.98, 'stress_factor': 0.9},
        'Wales': {'col_multiplier': 0.91, 'salary_premium': 0.94, 'stress_factor': 0.85},
        'Northern_Ireland': {'col_multiplier': 0.89, 'salary_premium': 0.93, 'stress_factor': 0.8}
    }

    data = []

    # Generate balanced behavior assignments
    n_behaviors = {behavior: int(n_records * pct) for behavior, pct in target_distribution.items()}
    remaining = n_records - sum(n_behaviors.values())
    n_behaviors['No_Change'] += remaining

    behavior_assignments = []
    for behavior, count in n_behaviors.items():
        behavior_assignments.extend([behavior] * count)
    np.random.shuffle(behavior_assignments)

    # ENHANCED: Create member archetypes for more realistic patterns
    member_archetypes = {
        'Young_Professional': {'age_range': (22, 32), 'education_boost': True, 'engagement_modifier': 0.8},
        'Mid_Career_Stable': {'age_range': (33, 45), 'education_boost': False, 'engagement_modifier': 1.1},
        'Senior_Career': {'age_range': (46, 58), 'education_boost': False, 'engagement_modifier': 1.3},
        'Pre_Retirement': {'age_range': (59, 67), 'education_boost': False, 'engagement_modifier': 1.5}
    }

    for i in range(n_records):
        member_id = f"NEST{str(i+100001).zfill(6)}"
        target_behavior = behavior_assignments[i]

        # Assign member archetype
        archetype = np.random.choice(list(member_archetypes.keys()),
                                   p=[0.25, 0.35, 0.30, 0.10])
        archetype_data = member_archetypes[archetype]

        # Generate demographics based on archetype
        age_min, age_max = archetype_data['age_range']
        age = np.random.uniform(age_min, age_max)

        gender = np.random.choice(['Male', 'Female'], p=[0.48, 0.52])

        # Enhanced education distribution by archetype
        if archetype == 'Young_Professional':
            education_probs = [0.10, 0.20, 0.45, 0.20, 0.05]
        elif archetype == 'Mid_Career_Stable':
            education_probs = [0.15, 0.25, 0.35, 0.20, 0.05]
        elif archetype == 'Senior_Career':
            education_probs = [0.25, 0.30, 0.25, 0.15, 0.05]
        else:  # Pre_Retirement
            education_probs = [0.35, 0.30, 0.20, 0.10, 0.05]

        education = np.random.choice(education_levels, p=education_probs)

        # BUSINESS LOGIC: Sector assignment with archetype influence
        if archetype == 'Young_Professional':
            sector_probs = [0.05, 0.15, 0.12, 0.10, 0.25, 0.18, 0.05, 0.05, 0.03, 0.02]
        elif education in ['Postgraduate', 'Professional']:
            sector_probs = [0.12, 0.10, 0.18, 0.15, 0.15, 0.20, 0.03, 0.03, 0.02, 0.02]
        else:
            sector_probs = [0.15, 0.15, 0.10, 0.08, 0.05, 0.05, 0.18, 0.12, 0.08, 0.04]

        sector = np.random.choice(sectors, p=sector_probs)
        region = np.random.choice(list(regional_factors.keys()))

        # ENHANCED: Multi-factor salary calculation
        salary_stats = sector_salary_stats[sector]
        base_salary = np.random.normal(salary_stats['mean'], salary_stats['std'])

        # Education multipliers (enhanced)
        education_multipliers = {
            'Secondary': 0.82, 'A_Level': 0.94, 'Undergraduate': 1.18,
            'Postgraduate': 1.42, 'Professional': 1.55
        }
        base_salary *= education_multipliers[education]

        # Experience curve (more sophisticated)
        experience_years = max(0, age - 18)
        if experience_years <= 5:
            experience_multiplier = 1 + (experience_years * 0.08)
        elif experience_years <= 15:
            experience_multiplier = 1.4 + ((experience_years - 5) * 0.03)
        else:
            experience_multiplier = 1.7 + ((experience_years - 15) * 0.015)

        base_salary *= experience_multiplier

        # Gender pay gap (sector-specific)
        if gender == 'Female':
            if sector in ['Technology', 'Finance']:
                base_salary *= 0.88  # Larger gap in male-dominated sectors
            else:
                base_salary *= 0.93

        # Regional adjustment
        regional_data = regional_factors[region]
        base_salary *= regional_data['salary_premium']

        # Target behavior influence on salary (subtle but important)
        if target_behavior == 'Increase' and np.random.random() < 0.3:
            base_salary *= 1.1  # Some high earners more likely to increase
        elif target_behavior == 'Pause_Stop' and np.random.random() < 0.4:
            base_salary *= 0.85  # Financial stress correlation

        # Final salary calculation
        base_salary += np.random.normal(0, salary_stats['std'] * 0.15)
        current_salary = max(salary_stats['min'],
                           min(salary_stats['max'], base_salary))

        # ENHANCED: Employment characteristics with business logic
        sector_tenure_months = {
            'Public': 60, 'Healthcare': 48, 'Education': 45, 'Finance': 36,
            'Manufacturing': 42, 'Technology': 28, 'Private_Services': 32,
            'Construction': 30, 'Retail': 24, 'Hospitality': 18
        }

        avg_tenure = sector_tenure_months[sector]
        if archetype in ['Senior_Career', 'Pre_Retirement']:
            avg_tenure *= 1.5

        tenure_months = max(3, np.random.exponential(avg_tenure))
        tenure_months = min(tenure_months, (age - 18) * 12)

        # Job changes with enhanced logic
        base_job_change_rate = 0.8
        if sector in ['Technology', 'Private_Services', 'Hospitality']:
            base_job_change_rate *= 1.4
        elif sector in ['Public', 'Healthcare', 'Education']:
            base_job_change_rate *= 0.6

        if archetype == 'Young_Professional':
            base_job_change_rate *= 1.6
        elif archetype in ['Senior_Career', 'Pre_Retirement']:
            base_job_change_rate *= 0.4

        job_changes_5yr = np.random.poisson(base_job_change_rate)
        years_with_nest = min(tenure_months / 12, np.random.exponential(4.2))

        # BUSINESS-CRITICAL: Enhanced financial situation modeling
        # Higher correlation with behavior patterns

        # Investment sophistication
        investment_sophistication = 0
        if education in ['Postgraduate', 'Professional']:
            investment_sophistication += 2
        elif education == 'Undergraduate':
            investment_sophistication += 1

        if current_salary > 60000:
            investment_sophistication += 2
        elif current_salary > 40000:
            investment_sophistication += 1

        if archetype in ['Senior_Career', 'Pre_Retirement']:
            investment_sophistication += 1

        has_other_pension = np.random.random() < min(0.85, 0.15 + investment_sophistication * 0.12)
        has_isa = np.random.random() < min(0.90, 0.25 + investment_sophistication * 0.15)

        # Advanced homeownership model
        homeowner_base_prob = 0.25 + (age - 25) * 0.012 + (current_salary - 25000) / 120000
        homeowner_base_prob *= regional_factors[region]['col_multiplier'] ** -0.5  # Inverse relationship
        homeowner = np.random.random() < min(0.95, max(0.1, homeowner_base_prob))

        # ENHANCED: Life events with behavior correlation
        recent_job_change_prob = 0.15
        if target_behavior in ['Decrease', 'Pause_Stop']:
            recent_job_change_prob *= 1.8  # Job instability correlates with contribution issues

        recent_job_change = np.random.random() < recent_job_change_prob

        # Salary changes with enhanced logic
        salary_change_prob = 0.25 if archetype == 'Young_Professional' else 0.18
        recent_salary_change = np.random.random() < salary_change_prob

        salary_change_pct = 0
        if recent_salary_change:
            if target_behavior == 'Increase':
                # Bias toward positive changes
                salary_change_pct = np.random.lognormal(np.log(0.08), 0.3) if np.random.random() < 0.8 else -np.random.lognormal(np.log(0.03), 0.2)
            elif target_behavior == 'Pause_Stop':
                # Bias toward negative changes
                salary_change_pct = -np.random.lognormal(np.log(0.05), 0.3) if np.random.random() < 0.7 else np.random.lognormal(np.log(0.04), 0.2)
            else:
                # Balanced changes
                salary_change_pct = np.random.lognormal(np.log(0.05), 0.3) if np.random.random() < 0.55 else -np.random.lognormal(np.log(0.04), 0.2)

        # Family and health events
        family_change_prob = {'Young_Professional': 0.18, 'Mid_Career_Stable': 0.12,
                             'Senior_Career': 0.08, 'Pre_Retirement': 0.06}[archetype]
        family_change = np.random.random() < family_change_prob

        health_event_prob = 0.02 + (age - 25) * 0.0012
        if target_behavior == 'Pause_Stop':
            health_event_prob *= 2  # Health events drive contribution pauses
        health_event = np.random.random() < health_event_prob

        # ENHANCED: Economic stress with regional and behavior correlation
        regional_stress_factor = regional_factors[region]['stress_factor']

        base_stress_probs = [0.30, 0.45, 0.25]  # Low, Medium, High

        # Adjust for target behavior
        if target_behavior in ['Decrease', 'Pause_Stop']:
            base_stress_probs = [0.15, 0.40, 0.45]  # Higher stress
        elif target_behavior == 'Increase':
            base_stress_probs = [0.50, 0.35, 0.15]  # Lower stress

        # Apply regional factor
        if regional_stress_factor > 1.1:
            base_stress_probs = [base_stress_probs[0] * 0.7, base_stress_probs[1], base_stress_probs[2] * 1.3]

        # Normalize probabilities
        base_stress_probs = np.array(base_stress_probs)
        base_stress_probs = base_stress_probs / base_stress_probs.sum()

        cost_of_living_stress = np.random.choice(['Low', 'Medium', 'High'], p=base_stress_probs)

        # Enhanced debt modeling
        debt_base = {'Young_Professional': 0.38, 'Mid_Career_Stable': 0.28,
                    'Senior_Career': 0.22, 'Pre_Retirement': 0.18}[archetype]

        if target_behavior in ['Decrease', 'Pause_Stop']:
            debt_base *= 1.4
        elif target_behavior == 'Increase':
            debt_base *= 0.7

        debt_to_income = max(0, np.random.normal(debt_base, 0.12))
        debt_to_income = min(0.75, debt_to_income)

        # BUSINESS-CRITICAL: Sophisticated disposable income calculation
        base_essential_pct = 0.62

        # Regional adjustment
        base_essential_pct *= regional_factors[region]['col_multiplier']

        # Family situation
        if family_change:
            base_essential_pct += 0.08

        # Stress level impact
        stress_multipliers = {'Low': 0.95, 'Medium': 1.0, 'High': 1.15}
        base_essential_pct *= stress_multipliers[cost_of_living_stress]

        # Homeownership impact
        if homeowner:
            base_essential_pct += 0.05  # Mortgage/maintenance costs
        else:
            base_essential_pct += 0.08  # Higher rent in many areas

        disposable_income = current_salary * (1 - base_essential_pct - debt_to_income)
        disposable_income = max(0, disposable_income)

        # ENHANCED: Contribution patterns with sector and archetype influence
        sector_contrib_base = {
            'Public': 0.095, 'Finance': 0.088, 'Technology': 0.085, 'Healthcare': 0.082,
            'Education': 0.090, 'Manufacturing': 0.075, 'Private_Services': 0.070,
            'Construction': 0.065, 'Retail': 0.058, 'Hospitality': 0.052
        }

        base_contrib = sector_contrib_base[sector]

        # Archetype adjustments
        archetype_contrib_multipliers = {
            'Young_Professional': 0.85, 'Mid_Career_Stable': 1.0,
            'Senior_Career': 1.15, 'Pre_Retirement': 1.35
        }
        base_contrib *= archetype_contrib_multipliers[archetype]

        # Education impact
        if education in ['Postgraduate', 'Professional']:
            base_contrib *= 1.2
        elif education == 'Undergraduate':
            base_contrib *= 1.1

        current_contribution_pct = max(0.01, np.random.normal(base_contrib, 0.022))
        monthly_contribution = current_salary / 12 * current_contribution_pct

        # ENHANCED: Engagement metrics with strong business correlation
        base_engagement = 25

        # Education impact
        education_engagement = {'Secondary': 0, 'A_Level': 5, 'Undergraduate': 12,
                              'Postgraduate': 20, 'Professional': 25}
        base_engagement += education_engagement[education]

        # Archetype impact
        base_engagement *= archetype_data['engagement_modifier']

        # Contribution level impact
        if current_contribution_pct > 0.10:
            base_engagement += 25
        elif current_contribution_pct > 0.08:
            base_engagement += 15
        elif current_contribution_pct > 0.06:
            base_engagement += 8

        # Investment sophistication impact
        base_engagement += investment_sophistication * 5

        # Target behavior correlation (key for model performance)
        if target_behavior == 'Increase':
            base_engagement *= 1.3
        elif target_behavior == 'No_Change':
            base_engagement *= 1.1
        elif target_behavior == 'Decrease':
            base_engagement *= 0.8
        else:  # Pause_Stop
            base_engagement *= 0.6

        # Generate individual engagement metrics
        app_logins_6m = max(0, np.random.poisson(base_engagement * 0.35))
        viewed_balance_6m = max(0, np.random.poisson(base_engagement * 0.55))
        opened_emails_6m = max(0, np.random.poisson(base_engagement * 0.28))
        contacted_support_6m = np.random.poisson(max(0.2, base_engagement * 0.02))

        engagement_score = min(100,
                              app_logins_6m * 2.8 +
                              viewed_balance_6m * 2.0 +
                              opened_emails_6m * 3.5 +
                              contacted_support_6m * 5.0)

        # ENHANCED: Knowledge and education metrics
        knowledge_base = {'Secondary': 20, 'A_Level': 28, 'Undergraduate': 38,
                         'Postgraduate': 48, 'Professional': 55}[education]

        # Pension calculator completion
        calc_prob = 0.12 + (education in ['Postgraduate', 'Professional']) * 0.25 + (archetype in ['Senior_Career', 'Pre_Retirement']) * 0.18
        if target_behavior == 'Increase':
            calc_prob *= 1.4
        elif target_behavior == 'Pause_Stop':
            calc_prob *= 0.5

        completed_pension_calc = np.random.random() < calc_prob

        # Webinar attendance
        webinar_prob = 0.06 + (education == 'Professional') * 0.12 + (archetype == 'Pre_Retirement') * 0.15
        if engagement_score > 60:
            webinar_prob *= 1.5
        attended_webinar = np.random.random() < webinar_prob

        # Calculate final knowledge score
        knowledge_score = knowledge_base
        if completed_pension_calc:
            knowledge_score += 25
        if attended_webinar:
            knowledge_score += 20
        if engagement_score > 60:
            knowledge_score += 12
        if has_other_pension:
            knowledge_score += 8

        knowledge_score = min(100, knowledge_score + np.random.randint(-8, 12))

        # Economic outlook with behavior correlation
        if target_behavior == 'Increase':
            outlook_probs = [0.40, 0.45, 0.15]  # More optimistic
        elif target_behavior == 'Pause_Stop':
            outlook_probs = [0.15, 0.35, 0.50]  # More pessimistic
        else:
            outlook_probs = [0.25, 0.50, 0.25]  # Balanced

        economic_outlook = np.random.choice(['Optimistic', 'Neutral', 'Pessimistic'], p=outlook_probs)
        inflation_impact = np.random.normal(0.058, 0.018)  # Current UK inflation environment

        # ADVANCED: Create financial stress and opportunity indices
        financial_stress_index = 0
        if debt_to_income > 0.4:
            financial_stress_index += 3
        elif debt_to_income > 0.3:
            financial_stress_index += 2
        elif debt_to_income > 0.2:
            financial_stress_index += 1

        if cost_of_living_stress == 'High':
            financial_stress_index += 3
        elif cost_of_living_stress == 'Medium':
            financial_stress_index += 1

        if disposable_income < 1500:
            financial_stress_index += 3
        elif disposable_income < 3000:
            financial_stress_index += 2
        elif disposable_income < 5000:
            financial_stress_index += 1

        if recent_job_change:
            financial_stress_index += 2
        if health_event:
            financial_stress_index += 2
        if family_change:
            financial_stress_index += 1

        # Opportunity index for increases
        opportunity_index = 0
        if disposable_income > 8000:
            opportunity_index += 3
        elif disposable_income > 5000:
            opportunity_index += 2
        elif disposable_income > 3000:
            opportunity_index += 1

        if recent_salary_change and salary_change_pct > 0.05:
            opportunity_index += 3
        elif recent_salary_change and salary_change_pct > 0.02:
            opportunity_index += 2

        if economic_outlook == 'Optimistic':
            opportunity_index += 2
        elif economic_outlook == 'Neutral':
            opportunity_index += 1

        if engagement_score > 70:
            opportunity_index += 2
        elif engagement_score > 50:
            opportunity_index += 1

        if knowledge_score > 70:
            opportunity_index += 2
        elif knowledge_score > 50:
            opportunity_index += 1

        # Engagement risk index
        engagement_risk_index = 0
        if app_logins_6m == 0:
            engagement_risk_index += 3
        elif app_logins_6m < 3:
            engagement_risk_index += 2
        elif app_logins_6m < 6:
            engagement_risk_index += 1

        if viewed_balance_6m < 2:
            engagement_risk_index += 2
        elif viewed_balance_6m < 5:
            engagement_risk_index += 1

        if opened_emails_6m == 0:
            engagement_risk_index += 2
        elif opened_emails_6m < 3:
            engagement_risk_index += 1

        if not completed_pension_calc and not attended_webinar:
            engagement_risk_index += 2

        # Create comprehensive record
        record = {
            'member_id': member_id,
            'age': round(age, 1),
            'member_archetype': archetype,
            'gender': gender,
            'education_level': education,
            'employment_sector': sector,
            'region': region,
            'current_salary': round(current_salary, 2),
            'tenure_months': round(tenure_months, 1),
            'years_with_nest': round(years_with_nest, 2),
            'job_changes_5yr': job_changes_5yr,
            'current_contribution_pct': round(current_contribution_pct, 4),
            'monthly_contribution': round(monthly_contribution, 2),
            'has_other_pension': has_other_pension,
            'has_isa': has_isa,
            'homeowner': homeowner,
            'investment_sophistication': investment_sophistication,
            'recent_job_change': recent_job_change,
            'recent_salary_change': recent_salary_change,
            'salary_change_pct': round(salary_change_pct, 4),
            'family_change': family_change,
            'health_event': health_event,
            'cost_of_living_stress': cost_of_living_stress,
            'debt_to_income': round(debt_to_income, 3),
            'disposable_income': round(disposable_income, 2),
            'app_logins_6m': app_logins_6m,
            'viewed_balance_6m': viewed_balance_6m,
            'opened_emails_6m': opened_emails_6m,
            'contacted_support_6m': contacted_support_6m,
            'engagement_score': round(engagement_score, 1),
            'completed_pension_calc': completed_pension_calc,
            'attended_webinar': attended_webinar,
            'knowledge_score': knowledge_score,
            'economic_outlook': economic_outlook,
            'inflation_impact': round(inflation_impact, 4),
            'financial_stress_index': financial_stress_index,
            'opportunity_index': opportunity_index,
            'engagement_risk_index': engagement_risk_index,
            'contribution_behavior': target_behavior  # TARGET VARIABLE
        }

        data.append(record)

    return pd.DataFrame(data)

# Generate business-optimized dataset
print("Generating business-optimized dataset with enhanced predictive features...")
df_business = generate_business_optimized_data(12000)

# Save to CSV
csv_filename = 'nest_business_optimized_dataset.csv'
df_business.to_csv(csv_filename, index=False)
print(f"Business-optimized dataset saved as '{csv_filename}'")

# Comprehensive analysis
print(f"Dataset shape: {df_business.shape}")
print(f"\n=== IMPROVED TARGET DISTRIBUTION ===")
behavior_counts = df_business['contribution_behavior'].value_counts()
print(behavior_counts)
print(f"\nBehavior percentages:")
for behavior, count in behavior_counts.items():
    percentage = (count / len(df_business)) * 100
    print(f"{behavior}: {count:,} members ({percentage:.1f}%)")

print(f"\n=== BUSINESS INTELLIGENCE INDICES ===")
print(f"Financial Stress Index: {df_business['financial_stress_index'].describe()}")
print(f"Opportunity Index: {df_business['opportunity_index'].describe()}")
print(f"Engagement Risk Index: {df_business['engagement_risk_index'].describe()}")

print(f"\n=== MEMBER ARCHETYPES ===")
archetype_behavior = pd.crosstab(df_business['member_archetype'],
                                df_business['contribution_behavior'],
                                normalize='index') * 100
print("Behavior distribution by archetype (%):")
print(archetype_behavior.round(1))

print(f"\n=== PREDICTIVE FEATURE CORRELATIONS ===")
# Calculate correlations with target variable
behavior_mapping = {'Pause_Stop': 0, 'Decrease': 1, 'No_Change': 2, 'Increase': 3}
df_business['behavior_numeric'] = df_business['contribution_behavior'].map(behavior_mapping)

key_features = ['financial_stress_index', 'opportunity_index', 'engagement_risk_index',
               'engagement_score', 'disposable_income', 'knowledge_score']

correlations = {}
for feature in key_features:
    corr = df_business[feature].corr(df_business['behavior_numeric'])
    correlations[feature] = corr
    print(f"{feature}: {corr:.3f}")

print(f"\n=== INTERVENTION TARGETING ===")
# High-priority segments for business action
high_stress = df_business['financial_stress_index'] >= 6
high_engagement_risk = df_business['engagement_risk_index'] >= 5
high_opportunity = df_business['opportunity_index'] >= 6

print(f"High financial stress members: {high_stress.sum():,} ({high_stress.mean()*100:.1f}%)")
print(f"High engagement risk members: {high_engagement_risk.sum():,} ({high_engagement_risk.mean()*100:.1f}%)")
print(f"High opportunity members: {high_opportunity.sum():,} ({high_opportunity.mean()*100:.1f}%)")

# Predictive power validation
pause_stop_members = df_business['contribution_behavior'] == 'Pause_Stop'
print(f"\nPause/Stop prediction validation:")
print(f"High stress correlation: {df_business[pause_stop_members]['financial_stress_index'].mean():.2f} vs {df_business[~pause_stop_members]['financial_stress_index'].mean():.2f}")
print(f"Low engagement correlation: {df_business[pause_stop_members]['engagement_score'].mean():.1f} vs {df_business[~pause_stop_members]['engagement_score'].mean():.1f}")

print(f"\n" + "="*70)
print("BUSINESS-OPTIMIZED DATASET COMPLETE")
print("KEY IMPROVEMENTS FOR BUSINESS RESULTS:")
print("✓ Balanced class distribution (40/30/20/10) for better model performance")
print("✓ Enhanced predictive indices (Financial Stress, Opportunity, Engagement Risk)")
print("✓ Member archetypes for targeted intervention strategies")
print("✓ Stronger feature correlations with business outcomes")
print("✓ Regional and demographic realism for scalable insights")
print("✓ Investment sophistication tracking for growth opportunities")
print("✓ Multi-dimensional engagement metrics for early warning systems")
print("="*70)
```
